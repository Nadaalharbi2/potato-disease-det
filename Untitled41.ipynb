```python
import os
import shutil
import random
import zipfile

# 1. حددي اسم ملفك المضغوط الموجود في Colab
zip_name = "/content/Potato_Dataset_2type_object_detection.zip"
extract_path = "/content/extracted_data"
base_out = "/content/potato_yolo_dataset"

print(" جاري فك ضغط الملف...")
with zipfile.ZipFile(zip_name, 'r') as zip_ref:
    zip_ref.extractall(extract_path)
print(" تم فك الضغط بنجاح!")

# 2. إنشاء هيكل مجلدات YOLO
for split in ['train', 'valid', 'test']:
    os.makedirs(f"{base_out}/images/{split}", exist_ok=True)
    os.makedirs(f"{base_out}/labels/{split}", exist_ok=True)

# 3. جمع كل أزواج الصور وملفات التسمية المقابلة لها
valid_exts = ('.jpg', '.jpeg', '.png', '.bmp', '.webp')
pairs = []

for root, _, files in os.walk(extract_path):
    for f in files:
        if f.lower().endswith(valid_exts):
            base_name = os.path.splitext(f)[0]
            txt_path = os.path.join(root, f"{base_name}.txt")
            if os.path.exists(txt_path):
                pairs.append((os.path.join(root, f), txt_path, f, f"{base_name}.txt"))

print(f" إجمالي الصور مع ملفات التسميات: {len(pairs)}")

# 4. خلط البيانات وتقسيمها
random.seed(42)
random.shuffle(pairs)

n_total = len(pairs)
n_train = int(n_total * 0.70)
n_val = int(n_total * 0.10)

splits = {
    'train': pairs[:n_train],
    'valid': pairs[n_train:n_train + n_val],
    'test': pairs[n_train + n_val:]
}

for split_name, items in splits.items():
    for img_src, txt_src, img_name, txt_name in items:
        shutil.copy(img_src, f"{base_out}/images/{split_name}/{img_name}")
        shutil.copy(txt_src, f"{base_out}/labels/{split_name}/{txt_name}")
    print(f" {split_name}: تم تجهيز {len(items)} صورة وتسمية")

# 5. إنشاء ملف التكوين data.yaml
yaml_content = f"""path: {base_out}
train: images/train
val: images/valid
test: images/test

names:
  0: healthy
  1: infection
"""

with open(f"{base_out}/data.yaml", "w") as f:
    f.write(yaml_content)

print("\  تجهيز مجلد التدريب وملف data.yaml بنجاح!")
```

```python
# تثبيت Ultralytics إذا لم تكن مثبتة
!pip install -q ultralytics

from ultralytics import YOLO

# تحميل أوزان النموذج الأساسي (خفيف وسريع للتدريب)
model = YOLO('yolov8n.pt')

# بدء التدريب
results = model.train(
    data='/content/potato_yolo_dataset/data.yaml',
    epochs=30,
    imgsz=640,
    batch=16,
    name='potato_detection_model'
)
```

```python
# تقييم النموذج وحساب المقاييس بالأرقام بدقة
metrics = model.val(data='/content/potato_yolo_dataset/data.yaml', split='test')

print("\n---   نتائج التقييم   ---")
print(f"Precision: {metrics.box.p.mean() * 100:.2f}%")
print(f"Recall:    {metrics.box.r.mean() * 100:.2f}%")
print(f"mAP@50:    {metrics.box.map50 * 100:.2f}%")
print(f"mAP@50-95: {metrics.box.map * 100:.2f}%")
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from ultralytics import YOLO

# 1. تحميل النموذج المدرب
model = YOLO('/content/runs/detect/potato_detection_model/weights/best.pt')


def estimate_disease_severity(image_path):
  img = cv2.imread(image_path)
  img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
  h, w, _ = img.shape

  # التنبؤ باستخدام YOLO
  results = model.predict(source=image_path, conf=0.25)

  severity_results = []

  for r in results:
    for box in r.boxes:
      cls_id = int(box.cls[0])
      cls_name = model.names[cls_id]

      # إذا كانت النتيجة مصابة
      if 'infect' in cls_name.lower() or 'affect' in cls_name.lower():
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        leaf_roi = img[y1:y2, x1:x2]

        # تحويل المنطقة المحددة إلى فضاء HSV
        hsv = cv2.cvtColor(leaf_roi, cv2.COLOR_BGR2HSV)

        # عزل مساحة الورقة
        mask_leaf = cv2.inRange(
            hsv, np.array([10, 30, 20]), np.array([90, 255, 255])
        )
        leaf_area = cv2.countNonZero(mask_leaf)

        # عزل البقع المتضررة
        mask_blight = cv2.inRange(
            hsv, np.array([10, 40, 20]), np.array([30, 255, 180])
        )
        blight_area = cv2.countNonZero(mask_blight)

        # حساب النسبة المئوية للشدة
        if leaf_area > 0:
          severity_pct = (blight_area / leaf_area) * 100
        else:
          severity_pct = 0.0

        # تحديد مستوى التدخل الميداني المقترح
        if severity_pct < 15:
          action = 'خفيف (Mild): إزالة الورقة يدوياً (Pinch off)'
        elif severity_pct <= 35:
          action = 'متوسط (Moderate): رش فطري موضعي (Targeted Spray)'
        else:
          action = 'شديد (Severe): بؤرة خطرة تستدعي العزل الفوري'

        severity_results.append((severity_pct, action))

        # رسم النتيجة على الصورة بحجم خط وسمك مناسبين للحجم الأصغر
        cv2.rectangle(img_rgb, (x1, y1), (x2, y2), (255, 0, 0), 2)
        cv2.putText(
            img_rgb,
            f'Severity: {severity_pct:.1f}%',
            (x1, max(y1 - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255, 0, 0),
            2,
        )

  # عرض الصورة بحجم أصغر وأنيق (4x4 بوصة)
  plt.figure(figsize=(4, 4))
  plt.imshow(img_rgb)
  plt.axis('off')
  plt.tight_layout()
  plt.show()

  return severity_results
```

```python
from google.colab import files

# نافذة لاختيار صورة من جهازك
print("ارفعي ورقة البطاطس لاختبار الشدة:")
uploaded = files.upload()

for filename in uploaded.keys():
    print(f"\nجاري تحليل الصورة: {filename}")
    results = estimate_disease_severity(filename)

    for idx, (sev, act) in enumerate(results, 1):
        print(f"الورقة {idx} -> نسبة الشدة: {sev:.2f}% | التوصية الميدانية: {act}")
```
